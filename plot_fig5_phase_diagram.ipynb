```python
import matplotlib.pyplot as plt  # plotting library
import numpy as np  # general math functions 
from matplotlib.colors import ListedColormap # colormap
from matplotlib import patches as mpatches # for legend

plt.style.use('aps.mplstyle')

# load data
data = np.load('data/fig5_phase_diagram.npz')
U2 = data['U2']      
U1 = data['U1']      
phase_data = data['p_values'] 

# plot the phase diagram
fig, ax = plt.subplots(figsize=(3.375,2.8))

# colors = ['#3B0F70', '#3E7989', '#2BB07F', '#FDE725']
colors = ['#8CC7A1', '#F09C81', '#7BA4CC', '#E8ECEF']
cmap = ListedColormap(colors)

labels_dict = {
    1: r"Topological II",
    2: r"Fragmentary",
    3: r"Topological I",
    4: r"Trivial"
}

mesh = ax.pcolormesh(U2, U1, phase_data, cmap=cmap, vmin=0.5, vmax=4.5, rasterized=True)

ax.set_xlabel(r"$U_2$")
ax.set_ylabel(r"$U_1$")

# ax.set_xlim(-3, 3)
# ax.set_ylim(0, 2)

legend_patches = [
    mpatches.Patch(color=colors[0], label=labels_dict[1]),
    mpatches.Patch(color=colors[1], label=labels_dict[2]),
    mpatches.Patch(color=colors[2], label=labels_dict[3]),
    mpatches.Patch(color=colors[3], label=labels_dict[4])
]

ax.legend(handles=legend_patches, loc='upper right', frameon=True)

plt.tight_layout()
plt.savefig('figures/fig5.pdf', dpi=300, bbox_inches='tight')
plt.show()
```
